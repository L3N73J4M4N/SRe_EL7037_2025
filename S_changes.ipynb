## Nusselt con dx y dy

```python
def get_SoH(row):
    r_eol = 0.12
    r_new = 0.032

    heats = []
    for c in list(row.index):
        if not pd.isna(row[c]) and c[0] == 'q':
            heats.append(row[c])
    heats = np.array(heats)
    arg_maxs = np.where(heats == heats.max())[0]

    if row['colIndex'] - 1 in arg_maxs:
        r = (heats.max() / heats.min()) * r_new
        row['SoH'] = (r_eol - r) / (r_eol - r_new)
    else:
        row['SoH'] = 1.0
    return row
```

```python
soh_25 = df_n_25.apply(get_SoH, axis=1)
soh_53 = df_n_53.apply(get_SoH, axis=1)
```

```python
df_25 = soh_25[['dx', 'dy', 'separation', 'Rem', 'prandtl', 'nusselt', 'SoH']].rename(
    {'separation': 'S', 'Rem': 'Re', 'prandtl': 'Pr', 'nusselt': 'Nu'}, axis=1
)
df_53 = soh_53[['dx', 'dy', 'separation', 'Rem', 'prandtl', 'nusselt', 'SoH']].rename(
    {'separation': 'S', 'Rem': 'Re', 'prandtl': 'Pr', 'nusselt': 'Nu'}, axis=1
)
```
